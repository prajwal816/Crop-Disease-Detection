```python
import pandas as pd
from pathlib import Path

MANIFEST_PATH = Path("/kaggle/input/datasets/katakuricharlotte/crop-csvfiles/plantvillage_selected_manifest.csv")
manifest = pd.read_csv(MANIFEST_PATH)

train_hashes = set(
    manifest.loc[manifest["split"] == "train", "sha256"].dropna()
)

val_hashes = set(
    manifest.loc[manifest["split"] == "val", "sha256"].dropna()
)

shared_hashes = train_hashes.intersection(val_hashes)

leakage_rows = (
    manifest[manifest["sha256"].isin(shared_hashes)]
    .sort_values(["sha256", "split", "image_path"])
)

leakage_path = Path("/kaggle/working/exact_duplicate_train_val_leakage.csv")
leakage_rows.to_csv(leakage_path, index=False)

print(f"Exact duplicate SHA-256 hashes across train and val: {len(shared_hashes)}")
print(f"Images involved in cross-split exact duplicates: {len(leakage_rows)}")
print(f"Saved: {leakage_path}")

if len(leakage_rows) == 0:
    print("PASS: No byte-identical train-val image leakage found.")
else:
    display(leakage_rows[[
        "split", "label", "file_name", "sha256", "image_path"
    ]].head(20))
```
